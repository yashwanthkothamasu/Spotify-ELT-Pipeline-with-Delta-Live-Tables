```python
df_DimTrack = spark.read.format("parquet")\
                   .load("abfss://bronze@adfyashwanthstorage.dfs.core.windows.net/DimTrack")
                   #Since, after stream is created due to errors I can't able to see the display.So, loading it again through batch data.
```

```python
display(df_DimTrack)
```

```python
df_DimTrack = df_DimTrack.withColumn("Duration_Flag", when(col("duration_sec") < 150, 'low').when(col("duration_sec") < 300, 'medium').otherwise('high'))
df_DimTrack = df_DimTrack.withColumn("track_name",regexp_replace(col("track_name"),'-',' '))
                                                    
display(df_DimTrack)
```

```python
from pyspark.sql.functions import initcap
df_DimTrack = df_DimTrack.withColumn("track_name",initcap(col("track_name")))
df_DimTrack = df_DimTrack.fillna({"album_name":"Unknown Album"}) # Filling Unknown Album at null values of album_name.fillna({},{},...)
df_DimTrack = df_DimTrack.filter(col("duration_sec") > 0)

display(df_DimTrack)
```

```python
df_DimTrack.write.format("delta")\
       .mode("append")\
       .option("path","abfss://silver@adfyashwanthstorage.dfs.core.windows.net/DimTrack/data")\
       .saveAsTable("spotify_cata.silver_v2.dimtrack")
```
